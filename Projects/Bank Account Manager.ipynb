```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.history = []

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit must be more than zero.")
        else:
            self.balance += amount
            self.history.append(f"Deposited {amount}")
            print("--- Deposit successful ---")
            print(f"Amount deposited: {amount}")
            print(f"New balance: {self.balance}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal must be more than zero.")
        elif amount > self.balance:
            print("You don't have enough money.")
            print(f"Current balance is {self.balance}.")
        else:
            self.balance -= amount
            self.history.append(f"Withdrew {amount}")
            print("--- Withdrawal successful ---")
            print(f"Amount withdrawn: {amount}")
            print(f"New balance: {self.balance}")

    def show_history(self):
        if len(self.history) == 0:
            print("No transactions yet.")
        else:
            for item in self.history:
                print(item)


FILE_NAME = "accounts.txt"


def load_accounts():
    accounts = {}
    try:
        file = open(FILE_NAME, "r")
        for line in file:
            line = line.strip()
            if line != "":
                name, password = line.split(",", 1)
                accounts[name] = password
        file.close()
    except FileNotFoundError:
        pass   
    return accounts


def save_account(name, password):
    file = open(FILE_NAME, "a")
    file.write(f"{name},{password}\n")
    file.close()


def login():
    accounts = load_accounts()
    name = input("Enter your account name: ").strip()

    if name == "" or "," in name:
        print("Invalid name. It can't be empty or contain a comma.")
        return None

    key = name.lower()

    if key in accounts:
        attempts = 3
        while attempts > 0:
            entered = input("Enter password: ")
            if entered == accounts[key]:
                print("Access granted.")
                return name
            attempts -= 1
            print(f"Wrong password. {attempts} attempt(s) left.")
        print("Too many wrong attempts. Goodbye!")
        return None

    print("New account detected. Let's create a password.")
    while True:
        password = input("Create a password: ")
        confirm = input("Confirm password: ")
        if password == "":
            print("Password can't be empty.")
        elif password != confirm:
            print("Passwords don't match. Try again.")
        else:
            break

    save_account(key, password)
    print("Account created and password saved!")
    return name


def main():
    print("Welcome to Namzoya's Bank Account Manager!")

    owner = login()
    if owner is None:
        return

    print(f"Hello, {owner}!")
    account = BankAccount(owner)

    while True:
        print("\n========Menu========")
        print("1. Deposit Money")
        print("2. Withdraw Money")
        print("3. Check Balance")
        print("4. Transaction History")
        print("5. Exit")

        choice = input("Please select an option (1-5): ")

        if choice == "1":
            try:
                amount = float(input("Enter the amount to deposit: "))
                account.deposit(amount)
            except ValueError:
                print("Invalid input. Please enter a valid number.")

        elif choice == "2":
            try:
                amount = float(input("Enter the amount to withdraw: "))
                account.withdraw(amount)
            except ValueError:
                print("Invalid input. Please enter a valid number.")

        elif choice == "3":
            print(f"Current balance for {account.owner}: {account.balance}")

        elif choice == "4":
            account.show_history()

        elif choice == "5":
            print("Thank you for using Namzoya's Bank Account Manager!")
            break

        else:
            print("Invalid option. Please select a valid option (1-5).")


main()
```
